```python
import pandas as pd

data= pd.read_csv('/content/creditcard.csv')
print("shape:",data.shape)
data.info()
print(data.head(5))
data.describe()
```

```python
print(data.isnull().sum())
```

```python
df=data.copy()
row,col=df.shape
print("rows:",row,"columns:",col)

#print(df['Class'].describe())
#class_1=df.loc[df['Class']==1]
#Fraud_transactions=len(class_1)

class_0,class_1=data["Class"].value_counts()
Fraud_transactions=class_1
print(class_1)
Non_fraud_transactions=row - Fraud_transactions
Fraud_percentage=(Fraud_transactions / row) * 100
non_fraud_percentage=(Non_fraud_transactions / row) * 100


print("Fraud_transactions:",Fraud_transactions,
      "\n Non_fraud_transactions:",Non_fraud_transactions,
      "\n Fraud_percentage:",Fraud_percentage,
      "\n non_fraud_percentage",non_fraud_percentage)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

stat=[Fraud_percentage,non_fraud_percentage]
classes=["fraud", "non fraud"]
#sns.countplot(df, x="Class")

fig=plt.figure(figsize=(8,5))
plt.pie(stat,labels=classes)

plt.title("distribution of classes")
plt.legend()
plt.show()
```
